```python
from anytree import Node, RenderTree

def get_subdict_by_path(d, path):
    """Navigate dict by path list, return subdict at that path."""
    subdict = d
    for p in path:
        subdict = subdict.get(p, {})
    return subdict

def dict_to_anytree_numbered(
    d, parent=None, depth=0, counters=None, show_files=True, max_depth=None, target_path=None
):
    if target_path:
        d = get_subdict_by_path(d, target_path)
    if counters is None:
        counters = {}
    if max_depth is not None and depth > max_depth:
        return
    if depth not in counters:
        counters[depth] = 1
    for key, value in d.items():
        if key == "Files":
            if show_files and (max_depth is None or depth <= max_depth):
                for file in value:
                    name = f"{counters[depth]}. {file}"
                    Node(name, parent=parent)
                    counters[depth] += 1
        else:
            name = f"{counters[depth]}. {key}"
            node = Node(name, parent=parent)
            counters[depth] += 1
            if isinstance(value, dict):
                dict_to_anytree_numbered(
                    value, node, depth + 1, counters.copy(), show_files=show_files, max_depth=max_depth
                )

# Example usage:
target_path = ["root", "subDir12"]  # Change this to your desired folder path
root_name = target_path[-1] if target_path else list(folderTreeDict.keys())[0]
root_node = Node(f"1. {root_name}")
dict_to_anytree_numbered(
    folderTreeDict, root_node, depth=1, show_files=True, max_depth=2, target_path=target_path
)

for pre, fill, node in RenderTree(root_node):
    print("%s%s" % (pre, node.name))
```
