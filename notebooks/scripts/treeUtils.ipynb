```python
from pathlib import Path
from xaidar.filesUtils import loadPickle

tree_dir = Path( "../../data/treeObjs/xchem/data")
treeFileName = "tree_2017_lb18145-35.pkl"
treeFile = tree_dir / treeFileName
tree = loadPickle( treeFile)
```

```python
from xaidar.treeObj import traceBackPath, cumulativeCount, findAllTargetIdxs, getGammaID, convertGammaIDtoFolderID, convertIDtoPath

def traceBackAllPaths(targetItem, tree, foldersCount, treeMaxDepth, startDepth = 1):
    try:
        lst_idxs = findAllTargetIdxs(targetItem, tree, startDepth = startDepth, endDepth= treeMaxDepth+2  )
    except:
        return None
    lst_paths = []
    lst_foldersID = []
    for idx in lst_idxs:
        supraFolderIdx, folderIdx, targetLevel = idx[0], idx[1], idx[2]
        folderSumCount = cumulativeCount( foldersCount )
        gammaFolderID = getGammaID(supraFolderIdx, folderIdx, targetLevel, folderSumCount )
        folderID = convertGammaIDtoFolderID( gammaFolderID, foldersCount )
        folderPath = convertIDtoPath( tree, foldersCount, folderID)
        lst_paths.append(folderPath)
        lst_foldersID.append(folderID)
    return lst_paths, lst_foldersID 

allPaths, _ = traceBackAllPath( "HAO1A-x0208", tree["fileTree"],  tree["foldersCount"], tree["folderTreeMaxDepth"] + 1 ) 
targetIdxs = traceBackPath( "HAO1A-x0208", tree["fileTree"],  tree["foldersCount"], tree["folderTreeMaxDepth"] + 1 ) #  "HAO1A-x0208"

print( targetIdxs )
print( allPaths )
```
