```python
import sys
import gc

def get_top_memory_objects(num_objects=20):
    """
    Retrieves a list of all objects currently in memory, sorts them by
    their memory consumption, and prints the top 'num_objects' in a formatted table.

    This is useful for quickly identifying potential memory leaks or unexpectedly large objects.
    Note: The size reported by sys.getsizeof() is the size of the object itself,
    not including the size of objects it references.
    
    Args:
        num_objects (int): The number of top objects to display.
    """
    # Force a garbage collection pass to ensure all objects are found
    gc.collect()

    # Get all objects tracked by the garbage collector
    all_objects = gc.get_objects()
    print(f"Total number of objects currently in memory: {len(all_objects)}")
    
    # Store objects and their sizes in a list of tuples (size, obj)
    sizes = []
    for obj in all_objects:
        try:
            sizes.append((sys.getsizeof(obj), obj))
        except TypeError:
            # Some objects, like custom C extensions, may not have a simple size
            continue

    # Sort the list by size in descending order
    sizes.sort(key=lambda x: x[0], reverse=True)

    # Print the top N largest objects
    print(f"\nTop {num_objects} memory-consuming objects:")
    print("-" * 80)
    print(f"{'Size (bytes)':<15} | {'Type':<30} | {'Value (truncated)':<30}")
    print("-" * 80)

    for size, obj in sizes[:num_objects]:
        # Truncate the value for clean printing
        value_str = str(obj).replace('\n', '\\n')
        if len(value_str) > 30:
            value_str = value_str[:27] + "..."
        
        # Use a `try-except` block to handle objects that can't be stringified
        try:
            type_name = type(obj).__name__
            print(f"{size:<15} | {type_name:<30} | {value_str:<30}")
        except:
            print(f"{size:<15} | {type_name:<30} | {'<unprintable>':<30}")
```

```python
from xaidar.filesUtils import roundBytes
def memory():
    """
    Get node total memory and memory usage
    """
    with open('/proc/meminfo', 'r') as mem:
        ret = {}
        tmp = 0
        for i in mem:
            sline = i.split()
            if str(sline[0]) == 'MemTotal:':
                ret['total'] = int(sline[1]) 
            elif str(sline[0]) in ('MemFree:', 'Buffers:', 'Cached:'):
                tmp += int(sline[1])
        ret['free'] = tmp 
        ret['used'] =  int(ret['total']) - int(ret['free']) 
        ret['used'] = roundBytes( ret['used']  )
        ret['free'] = roundBytes( ret['free']  )
        ret['total'] = roundBytes( ret['total'] )
    return ret
memory()
```

```python
def get_ram_info():
    """
    Retrieves and displays total and available RAM in a human-readable format.
    """
    mem = psutil.virtual_memory()

    # Convert bytes to gigabytes for readability
    total_gb = mem.total / (1024 ** 3)
    available_gb = mem.available / (1024 ** 3)
    used_gb = mem.used / (1024 ** 3)
    free_gb = mem.free / (1024 ** 3)
    percent_used = mem.percent

    print(f"Total RAM: {total_gb:.2f} GB")
    print(f"Available RAM: {available_gb:.2f} GB")
    print(f"Used RAM: {used_gb:.2f} GB")
    print(f"Free RAM: {free_gb:.2f} GB")
    print(f"Percentage Used: {percent_used:.2f}%")
    

# Call the function to display the information
get_ram_info()
```
