```python
from pathlib import Path
rootdir = Path("../../../..").resolve()
import sys
sys.path.insert(0, str(rootdir) )
from typing import Callable, Generator

import gemmi 
import parasail
import numpy as np
from rdkit import Chem

from xaidar.data.molecModels import loadPDB
from xaidar.data.molecModels import get_pdb_stats, sele_pdb, sele_Lig, get_res_CoM
from xaidar.data.molecModels import flatten_pdb, sele_AA, createPDB, get_atom_coord
```

```python
input_pdb_dir = rootdir.joinpath("data/md/00-input_pdb")
```

```python
ref_pdb = loadPDB( input_pdb_dir.joinpath("A0926a.pdb"))
```

```python
lig_CoM = flatten_pdb( sele_pdb( ref_pdb, sele_Lig ), "chain" )[0].calculate_center_of_mass()
print( lig_CoM)
```

```python
from xaidar.data.molecModels import sele_closest_Chain, createPDB, clear_empty, savePDB
prot_chain = sele_closest_Chain( flatten_pdb( sele_pdb( ref_pdb, sele_AA), "chain" ), lig_CoM ) 
print( prot_chain )
prot_chain[0].name = "A"
prot_pdb = createPDB( chainList = prot_chain )
clear_empty( prot_pdb)
get_pdb_stats( prot_pdb )
savePDB( prot_pdb, input_pdb_dir.joinpath( "A0926a_processed.pdb"))
savePDB( prot_pdb, input_pdb_dir.joinpath( "../01-gromacs_input/A0926a_processed.pdb"))
```
