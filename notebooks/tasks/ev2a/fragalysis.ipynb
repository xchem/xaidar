## Goal: Rearrage Files so that there is a non-duplicated folder with all the
## files from fragalysis related to EV2A Protease

#### Key Data Types:

- **pdb** | **pandda-eventmap** | **mtz** | **2fo-fc** | **fo-fc** | **sdf** | 
**cif** | 

Key Data Structures for each Data Type:

- Base: event map

- Sdf: **{dataset}_combined.sdf**

- pdb: **crystallographic_files/{dataset}/{datasubset}.pdb**

- mtz: **crystallographic_files/{dataset}/{datasubset}.mtz**

- 2fo-fc: **{datasubset}_sigmaa.ccp4** and 
**{datasubset}_sigmaa_crystallographic.ccp4** 

- fo-fc: **{datasubset}_diff.ccp4** and 
**{datasubset}_diff_crystallographic.ccp4** 

```python
# Set Root and Concat Dir variables
from pathlib import Path
rootDir = Path(r"D:\xaidar\data\my_fragalysis\ev2a")
dataset = "A71EV2A"

concatDir = Path( r"D:\xaidar\data\my_fragalysis\ev2a\concatDir")
concatDir.mkdir( parents = True, exist_ok = True )
print( concatDir.joinpath("./data").resolve())
```

```python
# Cp rootdir/map/* > concatdir/ 
from pathlib import Path
from shutil import copy2

def iterFileSys(rootDir:Path, relPath: str = "."):
    for item in rootDir.iterdir():
        if item.is_file():
            # print( item.as_posix() )
            # print(concatDir.joinpath( relPath).joinpath( item.name).resolve().as_posix() )
            copy2( item, concatDir.joinpath( relPath).joinpath( item.name) )
        elif item.is_dir():
            newPath = relPath +"/"+ item.name
            concatDir.joinpath( newPath).resolve().mkdir(parents=True,
                                                         exist_ok= True)
            iterFileSys( item, relPath = newPath )

iterFileSys( rootDir.joinpath( "map") )
```

```python
from pathlib import Path
def cpDirectory(srcDir: Path, targDir:Path, rootDir:Path, relPath: str = "."):
    for item in srcDir.iterdir():
        if item.is_file():
            # print( item.as_posix() )
            # print(concatDir.joinpath( relPath).joinpath( item.name).resolve().as_posix() )
            copy2( item, targDir.joinpath( relPath).joinpath( item.name) )
        elif item.is_dir():
            newPath = relPath +"/"+ item.name
            targDir.joinpath( newPath).resolve().mkdir(parents=True,
                                                        exist_ok= True)
            iterFileSys( item, relPath = newPath )

iterFileSys( rootDir.joinpath( "map") )
```

```python
# Add sdf extra files into it
from shutil import copy2

copy2( rootDir.joinpath("sdf/{}_combined.sdf".format(dataset)),
       concatDir.joinpath("{}_combined.sdf".format(dataset) ) )
```

```python
# Add pdb extra files into it
from shutil import copytree

copytree( src = rootDir.joinpath( "pdb/crystallographic_files"),
          dst = concatDir.joinpath("crystallographic_files") )
```

```python
# Add mtz extra files into it
from shutil import copy2

for dataset in rootDir.joinpath( "mtz/crystallographic_files").iterdir():
    newDatasetDir = concatDir.joinpath( "crystallographic_files/" \
                                        "{}".format(dataset.name) )
    newDatasetDir.mkdir( parents=True, exist_ok=True)
    for file in dataset.iterdir():
        newFile = newDatasetDir.joinpath( file.name )
        copy2( file, newFile)
```

```python
# Add 2fo-fc extra files into concat Dir

from shutil import copy2

for dataset in rootDir.joinpath( "2fofc/aligned_files").iterdir():
    newDatasetDir = concatDir.joinpath( "aligned_files/{}".format(dataset.name) )
    assert newDatasetDir.exists(), ("This"
        " dataset directory does not exist in: "
        "{}").format(newDatasetDir.as_posix() ) 
    # newDatasetDir.mkdir( parents=True, exist_ok=True)
    fileNameList = [ "{}{}".format(dataset.name, fileName) for fileName in 
                 ["_sigmaa.ccp4","_sigmaa_crystallographic.ccp4" ] ]
    for fileName in fileNameList :
        file = dataset.joinpath( fileName)
        assert file.exists(), "This file does not exist in rootdir/2fofc: "\
            "{}".format(file)
        newFile = newDatasetDir.joinpath( fileName )
        copy2( file, newFile)
```

```python
# Add fo-fc extra files into concat Dir


from shutil import copy2

for dataset in rootDir.joinpath( "fofc/aligned_files").iterdir():
    newDatasetDir = concatDir.joinpath( "aligned_files/" \
                                        "{}".format(dataset.name) )
    assert newDatasetDir.exists(), ("This"
        " dataset directory does not exist in: "
        "{}").format(newDatasetDir.as_posix() ) 
    # newDatasetDir.mkdir( parents=True, exist_ok=True)
    fileNameList = [ "{}{}".format(dataset.name, fileName) for fileName in 
                 ["_diff.ccp4","_diff_crystallographic.ccp4" ] ]
    for fileName in fileNameList :
        file = dataset.joinpath( fileName)
        assert file.exists(), "This file does not exist in rootdir/fofc: "\
            "{}".format(file)
        newFile = newDatasetDir.joinpath( fileName )
        copy2( file, newFile)
```
