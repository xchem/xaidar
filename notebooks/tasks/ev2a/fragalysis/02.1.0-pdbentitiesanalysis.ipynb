# **<p align="center"> PDB Entities Analysis </p>**

## <ins> Description </ins>
### There are several main data formats (e.g. `ccp4`, `pdb`, `sdf`, ...). Here we do the most basic analysis of the pdb formats, by looking at the type of molecule, chain, protein and other entities that are contained in the pdbs.

<details open>
<summary> <font size="6"> <ins>Main Stpes </ins> </font> </summary>

- All PDBs and prtein-wide analysis

    - Extract all the different protein chains from all the different pdb formats

    - Keep a metadata pandas object, that links each chain to the specific `folder type`, `dataset`, `subset`, `file format`, and `chain id`.

    - Main Data analysis:

        - What type of entities exist? 
        
            - What are all the residue labels that exist?

            - How are the ligands labelled?
        
        - How many protein chains per pdb?

        - Is there at least one ligand per pdb?

        - Are there duplicate entities within a pdb?

        - Do the entities have alternative conformations (`altconf`)?
