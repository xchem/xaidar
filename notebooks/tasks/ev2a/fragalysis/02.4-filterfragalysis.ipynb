    - In crystallography:  structural context = Biological Assembly (biological binding context) + Crystallographic Artefacts

- the next step: identify which **compounds** have **protein contexts** which can be aligned to one another well

### <ins> Column Labels Individual Analysis </ins>

#### Analyse `'Code'` Column Regex Content

```python
import re
lst_regex_pattern = ["^A[0-9]*[a-z]$" ] # E.g. A0450a
lst_regex_obj = [ re.compile(pattern) for pattern in lst_regex_pattern]

regex_match_dict = { "match": [], "mismatch": []}


for value in df['Code'].values:
    matched = False
    for regex_obj in lst_regex_obj: 
        if regex_obj.search( value): 
            regex_match_dict["match"].append( value )
            matched = True
    if not matched: regex_match_dict["mismatch"].append(value)

print(regex_match_dict["mismatch"])
if regex_match_dict["mismatch"].__len__() == 0:
    print( "\'Code\' Label is composed of the following regex:\n\t{}" \
    "".format(lst_regex_pattern))

print("\nOne Dataset can have multiple letter endings (e.g. a, b, c), " \
"\neach representing a different protein site where the ligand was found" \
"\nwithin the same single crystal, indicating binding at multiple sites")
```

#### Create and Populate `fragscreen`, `piperid`, and `nonactive` folders

```python
# Copy Key Folders and files into new directories
from shutil import copy2

srcRootDir = Path( "./../../../../data/ev2a/fragalysis/01-concatDir/aligned_files" ).resolve()

def cp_codes( srcDir, destDir, codes ):
    for code in codes:
        dest = destDir / code
        dest.mkdir( exist_ok=True, parents=True )
        src = srcDir / code
        if not src.exists():
            print( f"Source directory not found: {src}, skipping..." )
            continue
        for srcFile in src.iterdir():
            filename = srcFile.name
            dstFile =  dest / filename
            copy2( srcFile, dstFile)
    print( f"Copied {len(codes)} codes to {destDir}" )

# cp_codes( srcRootDir, fragdir, fragScreen_codes ) # Copy Fragments into New Folder
# cp_codes( srcRootDir, piperidir, pipedirine_codes ) # Copy Piperidines into New Folder
# cp_codes( srcRootDir, nonactivedir, nonactivesite_codes ) # Copy Non-actives into New Folder
```

### Merge Subset Ligand Files (`piperid_lig`, `fragscreen_lig`, `startchem_lig`)

```python
# Merge Ligands into single file for ease of use in PyMOL
from rdkit import Chem

def merge_sdf_files( input_files, output_file ):
    writer = Chem.SDWriter(output_file)
    for filename in input_files:
        try:
            supplier = Chem.SDMolSupplier(filename)
        except FileNotFoundError:
            print(f"File not found: {filename}")
            continue
        for mol in supplier:
            if mol is not None:
                writer.write(mol)
    writer.close()
    print(f"Successfully written merged file to: {output_file}")

# input_files = [ str(piperidir / code / f"{code}.sdf") for code in pipedirine_codes ]
# output_file = rootdir / 'piperidir_lig.sdf'
# merge_sdf_files( input_files, output_file )
# input_files = [ str(fragdir / code / f"{code}.sdf") for code in fragScreen_codes ]
# output_file = rootdir / 'fragscreen_lig.sdf'
# merge_sdf_files( input_files, output_file )

# copy2( piperidir / "A0926a"/ "A0926a.sdf", rootdir / "startchem_lig.sdf" )
```
