```python
from plip.structure.preparation import PDBComplex

my_mol = PDBComplex()
my_mol.load_pdb('/home/eoo22534/mydir/xaidar/data/ev2a/fragalysis/02-chemofint/reference/A0926a.pdb') # Load the PDB file into PLIP class
print(my_mol) # Shows name of structure and ligand binding sites
my_bsid = 'LIG:A:250' # Unique binding site identifier (HetID:Chain:Position)
my_mol.analyze()
my_interactions = my_mol.interaction_sets[my_bsid] # Contains all interaction data

# Now print numbers of all residues taking part in pi-stacking
print([pistack.resnr for pistack in my_interactions.pistacking]) # Prints [84, 129]
# print([pistack.resnr for pistack in my_interactions.picat])  # Prints [84, 129, 165]
```

```python
from plip.basic import config
all_interactions = []


# Iterate over all detected binding sites
for binding_site in my_mol.interaction_sets:
    
    # Iterate over all possible interaction types defined by PLIP
    for interaction_type in config.interaction_type:
        
        # Get the list of specific interactions of that type (e.g., binding_site.hbonds)
        interactions = getattr(binding_site, interaction_type)
        
        for interaction in interactions:
            
            # Extract common details (varies slightly by interaction type)
            record = {
                "Binding_Site_ID": binding_site.site_id,
                "Interaction_Type": interaction_type,
                "Ligand_PDB_ID": interaction.ligand.pdbid,
                "Residue_Name": interaction.resnr, # Protein residue number
                "Residue_Chain": interaction.reschain,
                # Additional details like distance, specific atom names, etc.
                "Distance": interaction.distance if hasattr(interaction, 'distance') else 'N/A'
            }
            all_interactions.append(record)
```
