```python
# Check if the common piperidine code between cdd and fragalysis have the same SMILES
piperidine_frag_cdd_matching_codes = sorted( list( set( piperidine_frag_cdd_matching_codes) ) )

cdd_smiles = []
fraglys_smiles = []

for code in piperidine_frag_cdd_matching_codes:
    cdd_smiles.append( cdd_df[ cdd_df["Molecule Name"] == code ]["SMILES"].to_list() )
    regex = re.compile( f"^{code}.*" )
    fraglys_smiles.append( list( set(piperidine_df[ [ bool( regex.match( str(val)) ) for val in piperidine_df["Compound code"] ] ]["Smiles"].to_list() ) )  )

print( "Number of matched smiles per code in CDD: ", set([ len(smiles) for smiles in cdd_smiles ]) )
set_fraglys_num_smiles = set([ len(smiles) for smiles in fraglys_smiles ])
print( "Number of smiles per code in FRAGALYSIS: ", set_fraglys_num_smiles )
dict_fraglys_num_smiles = { num_smiles : [ (idx, smiles) for idx, smiles in enumerate( fraglys_smiles) if len(smiles) == num_smiles ]  for num_smiles in set_fraglys_num_smiles }
print( "Distribution of number of smiles per code in FRAGALYSIS: ",)
for num_smiles in dict_fraglys_num_smiles.keys():
    print( f"  Number of SMILES: {num_smiles} \t Number of Codes: {len( dict_fraglys_num_smiles[num_smiles] ) }" )

lst_match_code_and_smiles = []

for idx, (cdd_smile, fraglys_smile) in enumerate( zip(cdd_smiles, fraglys_smiles)):
    if len(fraglys_smile) > 1: continue
    if cdd_smile[0] == fraglys_smile[0]:
        lst_match_code_and_smiles.append( piperidine_frag_cdd_matching_codes[idx] )

print( "\nTotal Number of codes with matching smiles in Fragalysis and CDD: ", len(lst_match_code_and_smiles) )
print("Number of unique codes with matching smiles in Fragalysis and CDD: ", len(set(lst_match_code_and_smiles) ) )
print()

print( "######## COMPARING SMILES BETWEEN CDD AND FRAGALYSIS FOR COMMON PIPERIDINE CODES #############" )
mismatched_smiles = []
for idx, code in  enumerate( piperidine_frag_cdd_matching_codes ):
    cdd_smile_set = set( cdd_smiles[idx] )
    fragalysis_smile_set = set( fraglys_smiles[idx] )
    if cdd_smile_set != fragalysis_smile_set:
        mismatched_smiles.append( ( idx, code, cdd_smile_set, fragalysis_smile_set) )
        # print( f"Code: {code} \n CDD SMILES: {cdd_smile_set} \n FRAGALYSIS SMILES: {fragalysis_smile_set} \n" )
print( "Total Number of Piperidine Codes with Mismatched SMILES between CDD and FRAGALYSIS: ", len( mismatched_smiles ) )
print()
frag_one_smile_cdd_mismatch = [ ( idx, code, cdd_smile_set, fragalysis_smile_set) for ( idx, code, cdd_smile_set, fragalysis_smile_set) in mismatched_smiles if len( fragalysis_smile_set) == 1 ]
print( "Number of Piperidine Codes with Mismatched SMILES between CDD and FRAGALYSIS where FRAGALYSIS has only one SMILE: ", len( frag_one_smile_cdd_mismatch ) )
```
