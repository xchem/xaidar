```python
from pathlib import Path
import sys
sys.path.insert(0, Path( "../../.."))
```

```python
# Count Number of files and size for xchem : data/  and xchem : dataset/

from xaidar.filesUtils import loadPickle

xchemRawSizesDirPath = Path("../../../data/s3Sizes/xchem/raw")
rootDirs = ["data", "dataset"]
content = { "dataset": { "filesCount": 0, "rootDirSize": 0}, "data": {"filesCount": 0, "rootDirSize": 0}}

for fragPath in xchemRawSizesDirPath.iterdir():
    if fragPath.is_file():
        frag = loadPickle(fragPath)
        for dir in rootDirs:
            lst_sizes = [ value for key, value in frag.items() if key.split("/")[0] == dir and  type(value) == int ]
            content[dir]["filesCount"] += len(lst_sizes)
            content[dir]["rootDirSize"] += sum(lst_sizes)


print( "Done")
# Estimated Time: 4 m 31
```

```python
from xaidar.filesUtils import roundBytes
for dir in rootDirs:
    storSize, fileCount = content[dir]["rootDirSize"], content[dir]["filesCount"]
    print(f"Statistics for xchem : {dir}/")
    print( f"\tStorage Size: {roundBytes(storSize)}, Files Count: {fileCount}" )
```
