```python
import re

def countMatches(  lst, regex_pattern):
    count = len( ["" for item in lst if re.search( regex_pattern, item)])
    return count 

def listMatches( lst, regex_pattern):
    return  [ item for item in lst if re.search( regex_pattern, item)]

def findMatch(lst, regex_pattern, foo):
    match = foo( lst, regex_pattern)
    return match 

def findAllMatches( lst, lst_regex_patterns, lst_names, foo):
    """
    lst_names: name to attribute to each regex pattern match. One per regex
    """
    output_dict = {}
    for regex_pattern, key in zip( lst_regex_patterns, lst_names):
        output_dict[key] = findMatch( lst, regex_pattern, foo)
    return output_dict

lst_regex_patterns = [ f"/{weirdDir}/[a-z][a-z][0-9]+-[0-9]+$" for weirdDir in lst_weirdDirs ]

count = findAllMatches( lstSessionDirs, lst_regex_patterns, lst_weirdDirs, countMatches)
lst_files = findAllMatches( lstSessionDirs, lst_regex_patterns, lst_weirdDirs, listMatches)

for key, value in count.items():
    print( f"{key}: {value}")

print("\nList of Matching Directories")
    
for key, value in lst_files.items():
    print( f"\t{key}: {value}")
```
