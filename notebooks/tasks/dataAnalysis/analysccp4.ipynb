```python
# Create a Small Grid 
import gemmi as g
import numpy as np

np.random.seed(42)  # For reproducibility

grid = g.FloatGrid(1, 10, 10)
print("Grid Dimesions:", grid.nu, grid.nv, grid.nw )

# Fill the grid with random values
rand_vals = np.random.rand(grid.nu, grid.nv, grid.nw).round(2)
print(rand_vals[0, :, : ])

for i in range(grid.nu):
    for j in range(grid.nv):
        for k in range(grid.nw):
            grid.set_value(i, j, k, rand_vals[i, j, k])

# Print the grid values
print("Grid Values:")
matrix = grid.array
print(matrix[0, :, :])
```
