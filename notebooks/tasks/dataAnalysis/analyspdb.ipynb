#### Chains and Subchains:

["Polymeric chains [are] made of covalently linked building blocks, such as amino acids and nucleotides"](https://www.rcsb.org/docs/general-help/identifiers-in-pdb#entry-level-identifiers-)

Two sets of chain IDs are found in each PDB entry:
- one assigned by the PDB (`label_asym_id`), usually beginning with the alphabet A
- the other selected by the author (`auth_asym_id`) at the time of deposition.

Gemmi Distinguished between the two as follows:
- Chain Name = `auth_asym_id`
- Subchain Name = `label_asym_id`

Although in the PDB files they can have the same or different names, in Gemmi they always have different names according to the following formula:

&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;&emsp;`subchain Name = f"{chain.name}x{part of chain identifier}"`

Model contains chains (class `Chain`) that can be accessed either by index or by name. They can be considered "parallel identifiers".

"The polymer (residues before the TER record in the PDB format) goes into one subchain; all the other (non-polymer) residues are put into single-residue subchains; except the waters, which are all put into one subchain"
