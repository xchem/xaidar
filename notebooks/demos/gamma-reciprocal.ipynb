# <p align = "center"> Performing the Opposite Operation </p>

Given a folder or file ID, the task is to obtain the Gamma for the right level.
Now we have the reverse task. Given a gamma we want to get the file or folder ID.

In this case, we will be using the following objects:
- **Cumulative Count**: 
    - I.e. 
        - FoldersCount = [ [1], [3] , [5, 1, 2], [0, 0, 1, 0, 0, 0, 2, 1], [0, 0, 0, 0] ]
        - CumulativeCount = [ [1], [3], [5, 6, 8], [0, 0, 1, 1, 1, 1, 3, 4], 
    [0, 0, 0, 0] ]
- **GammaID**:     GammaID would be similar to folder ID, but each element represents the gamma for the correspondent level to select the right folder.
    - I.e. Given the top Folder and Cumulative Counts
        - Folder ID = [0, 2, 0, 1 ]
        - Gamma ID =  [ ] 

Example:
- File ID = [0, 2, 0, 18]
- Folder ID = File ID[:-1] = [0, 2, 0 ]
- FoldersCount = [ [1], [3] , [5, 1, 2], [0, 0, 1, 0, 0, 0, 2, 1], [0, 0, 0, 0] ]

Therefore, if we go down the folders count with the folder ID we get:
- Gamma = 7

Now, let's do the opposite:
- File Index = 18
- Gamma = 6
- File Level = 4,
