### <p align="center"> Mimicking File System Tree Hierarchies as Lists of Lists </p>

```python
from pathlib import Path
import sys
sys.path.insert(0, Path("../..").resolve().absolute().__str__() )
```

#### Convert list of Paths into Nested List of Lists

```python
# Import list of paths
from xaidar.filesUtils import loadPickle
from xaidar.treeObj import sortPaths, createFolderTree, pinchLevel, countFolders, createFileTree, displayTree, visualizeTree, createTree

objKeysPath = Path( r"..\..\data\s3ObjKeys\xchem\perProj\data\2017_lb13385-109.pkl") 
objKeys = loadPickle( objKeysPath)
print("I.e. of Raw object Keys:")
for path in objKeys[:3]: print( "\t- ",path )

## Step-by-Step Approach
# Order paths in alphabetical order, and truncate into folder only paths
orderedfolderPaths = sortPaths( objKeys )
print("\nI.e. of Sorted object Keys:")
for path in orderedfolderPaths[:3]: print( "\t- ",path )

# Create a nested list of list for all the existent folders
folderTree, treeDepth, folderIDs = createFolderTree( orderedfolderPaths )
print("\nI.e. of Folder Tree:")
displayTree( folderTree, 6)
visualizeTree( folderTree,   minMaxDepth=(0,6) )

# Create a list representing the number of Subfolders in the nested list of lists
foldersCount = countFolders( folderTree, treeDepth )
print("\nI.e. of SUB-folders Count Per Level:")
for level, count in enumerate( foldersCount[:6] ): print( f"\t- Level {level+1} : {count} ")
    # This means that Level 6 has 23 subfolderd divided by its 8 folders   

# Add files to tree with only folders
fileTree = createFileTree( objKeys, orderedfolderPaths, folderTree, folderIDs, foldersCount)
print("\nI.e. of File Tree:")
displayTree( fileTree, 6 )
visualizeTree( fileTree, minMaxDepth=(0, 6))

# ## OR All-at-Once Approach
# fileTree, folderTree, folderTreeMaxDepth,  foldersIDS, foldersCount, orderedFolderLst = createTree( objKeys )

```

#### Browse Through the Tree Like a File System

- Browse Through File ID 

```python
from xaidar.treeObj import viewSubtree

# Keep adding to folder ID
folderID = [0]

viewSubtree( fileTree, foldersCount, 0, len(folderID) + 2, folderID = folderID)
```

- Browse Through File Path

```python

# Keep adding to folder Path 
folderPath = "data"
viewSubtree( fileTree, foldersCount, 0, len(folderID) + 2, folderPath = folderPath)
```

```python
from xaidar.treeObj import openFolder

folderID = [0, 0, 0, 1, 1, 1, 1, 2, 0, 0]
folderContent = openFolder( fileTree, foldersCount, folderID)

print( f"The folder contains the following items:\n\t{folderContent}")
```
