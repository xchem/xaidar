# RDKit Tools Extensions

- ## rdkit_to_gemmi

```python
# Load modules and test data

import gemmi
from rdkit import Chem

caffeine_sdf = Chem.MolFromMolFile( str( Path("../../../../tests/testdata/molec/caffeine.sdf").resolve() ) )
```

```python
def rdkit_to_gemmi( rdkit_mol_lst: list[Chem.Mol] ):
    """
    Convert RDKit molecule to Gemmi structure.
    Args:
    - rdkit_mol: RDKit molecule object
    return:
    - gemmi_struct: Gemmi structure object
    """
    lst_gemmi_struct = []
    for rdkit_mol in rdkit_mol_lst:
        if rdkit_mol is None:
            continue
        pdb_block = Chem.MolToPDBBlock( rdkit_mol )
        gemmi_struct = gemmi.read_pdb_string( pdb_block )
        lst_gemmi_struct.append( gemmi_struct )
    return gemmi_struct
```

```python
from xaidar.data.molecModels import get_pdb_stats
gemmi_struct = rdkit_to_gemmi( [caffeine_sdf] )
pdb_stats = get_pdb_stats( gemmi_struct )
```
